```python
# Defining functions

def least_difference(a, b, c): 
# def means I am defining a function
# least_difference - name of the function
# (a, b, c) - Parameters / variable inside the function
 
    """Return the smallest difference between any two numbers 
    among a, b and c.
    """
    # DOCSTRING - explains what the function does
    #             also for docuemntation help()
                 
    # Function Body #
    
    # abs() takes the absolute value of the number 
    diff1 = abs(a - b) # (5 - 6) = 1
    diff2 = abs(b - c) # (6 - 7) = 1
    diff3 = abs(a - c) # (5 - 7) = 2
    return min(diff1, diff2, diff3) # min() gets the smallest value
    # return 1
    
print(
 least_difference(5, 6, 7), # Python allows trailing commas in argument lists. 
)
# Output: 1
```

```python
# Functions that don't return 

def least_difference(a, b, c):
    """Return the smallest difference between any two numbers
    among a, b and c.
    """
    diff1 = abs(a - b)
    diff2 = abs(b - c)
    diff3 = abs(a - c)
    min(diff1, diff2, diff3)
    
print(
    least_difference(5, 6, 7),
)

# Outputs to None
# similar to the concept of "null" in other languages.
```

```python
#Functions applied to other functions

def mult_by_five(x): 
    return 5 * x # 5 * 1 = 5

def call(fn, arg):  #fn - function, arg - value
    """Call fn on arg"""
    return fn(arg)  # call the function stored in fn using arg
                    # becomes mult_by_five(1)

def squared_call(fn, arg): 
    """Call fn on the result of calling fn on arg"""
    return fn(fn(arg)) 

# 1. squared_call(mult_by_five, 1)
#
# 2. This becomes mult_by_five(mult_by_five(1))
#
# 3. mult_by_five returns 5 becomes mult_by_five(5)
#
# 4. calls mult_by_five again them becomes 25

print(
    call(mult_by_five, 1),
    squared_call(mult_by_five, 1), 
    sep='\n', # '\n' is the newline character - it starts a new line
)
```

```python
#Higher-order function takes another function as an argument or returns a function as a result
#It is a function that works with other functions—either by taking them in or returning them.

def mod_5(x):
    """Return the remainder of x after dividing by 5"""
    return x % 5

print(
    'Which number is biggest?',
    max(100, 51, 14),
    'Which number is the biggest modulo 5?',
    max(100, 51, 14, key=mod_5), # mod_5(100), mod_5(51), mod_5(14)
    # 100 % 5 = 0
    # 51 % 5 = 1
    # 14 % 5 = 4
    sep='\n',
)
```
