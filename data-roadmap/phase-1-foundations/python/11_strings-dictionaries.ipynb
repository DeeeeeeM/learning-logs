Strings and Dictionaries

Strings in Python can be defined using either single or double quotations. They are functionally equivalent.


```python
x = 'Pluto is a planet'
y = "Pluto is a planet"
print(x == y)
```

Double quotes are convenient if your string contains a single quote character

It's also easy to create a string that contains double-quotes if you wrap it in single quotes:

```python
print("Pluto's a planet!")
print('My dog is named "Pluto"')
```

If we use the same type of quote, Python gets confused, to fix that we can use an escape "/"

```python
print('Pluto\'s a planet!')
```

What you type   What you get    Example                     print(example)
\'              '               'What\'s up?'               What's up?
\"              "               "That's \"cool\""           That's "cool"
\\              \               "Look, a mountain: /\\"     Look, a mountain: /\
\n              new line        "1\n2 3"                    1
                                                            2 3

Python's triple quote syntax for strings lets us include newlines literally (i.e. by just hitting 'Enter' on our keyboard, rather than using the special '\n' sequence). 

```python
triplequoted_hello = """hello
world"""
print(triplequoted_hello)
```

The print() function automatically adds a newline character unless we specify a value for the keyword argument end other than the default value of '\n':

```python
print("hello")
print("world")
print("hello", end='')
print("pluto", end='')
```

Strings are sequences¶
Almost everything we can do to a list is also possible to a string.

```python
planet = 'Pluto'

#length
print(len(planet))

#indexing
planet[0]     # E
planet[-1]    # h

#we can even loop over them
[char + '! ' for char in planet]
```

However, strings are immutable, so we cannot change them using the typical appends or assigns

```python
planet[0] = 'B'
planet.append("A")
```

str has lots of very useful methods.

```python
claim = "Pluto is a planet!"
claim.upper()
claim.lower()
# Searching for the first index of a substring
claim.index('plan')
claim.startswith(planet)
claim.endswith('planet')
```

```python
#str.split() turns a string into a list of smaller strings, breaking on whitespace by default. 

datestr = '1956-01-31'
year, month, day = datestr.split('-')
print (year + month + day)
```

```python
# str.join() sews a list of strings up into one long string, using the string it was called on as a separator.

'/'.join([month, day, year])
```

```python
#also possible for unicode emojis
words = ['PLUTO', 'IS', 'A', 'PLANET']
' 👏 '.join([word.upper() for word in words])
```

If we want to throw in any non-string objects, we have to be careful to call str()

```python
position = 9
planet + ", you'll always be the " + str(position) + "th planet to me."
```

```python
#To avoid confusion, just use str.format().
"{}, you'll always be the {}th planet to me.".format(planet, position)
```

```python
pluto_mass = 1.303 * 10**22
earth_mass = 5.9722 * 10**24
population = 52910390
# 2 decimal points / 3 decimal points / format as percent / separate with commas
"{} weighs about {:.2} kilograms ({:.3%} of Earth's mass). It is home to {:,} Plutonians.".format(
    planet, pluto_mass, pluto_mass / earth_mass, population,
)
```

Dictionaries built-in Python data structure for mapping keys to values.

```python
numbers = {'one':1, 'two':2, 'three':3}
#'one', 'two', and 'three' are the keys, and 1, 2 and 3 are their corresponding values.
```

Values are accessed via [] similar to indexing

```python
numbers['one']
```

```python
#We can use the same syntax to add another key

numbers['eleven'] = 11
numbers

# Or change the value with an existing key
numbers['three'] = 'Pluto'
numbers
```

Python has dictionary comprehensions with a syntax similar to the list comprehensions

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
planet_to_initial = {planet: planet[0] for planet in planets}
planet_to_initial
```

The in operator tells us whether something is a key in the dictionary

```python
'Saturn' in planet_to_initial
```

A for loop over a dictionary will loop over its keys


```python
for k in numbers:
    print("{} = {}".format(k, numbers[k]))
```

We can access a collection of all the keys or all the values with 
dict.keys() and dict.values().

```python
# Get all the initials, sort them alphabetically, and put them in a space-separated string.
' '.join(sorted(planet_to_initial.values()))
```

```python
' '.join(sorted(planet_to_initial.keys()))
```
