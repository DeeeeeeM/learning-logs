# In Python, everything is a “thing” that carries information with it.

"x = 12"

A number is not just 12 sitting there.

**Python** treats numbers in a unified way.

“Even normal numbers can be seen as complex numbers.”

It’s more like a little box that contains:
- The value 12
- plus some built-in extras Python attached to it



Everything in Python is a thing (object)

Objects carry:
- attributes → information (x.imag)
- methods → actions (x.bit_length())

What is (.) for?
It means “Give me something that is inside the box.”

Think of an object like a smart phone. It has:
- data (things it stores) → attributes
- actions it can do → methods

```python
x = 12 # Object

# A. Attribute (just information)
x.imag
# Means: “What is stored as the imaginary part?”
# there is no imaginary component, so its output is 0 

# B. Method (an action)
# Has parentheses () because it does something.
x.bit_length() # gets the length of bits it has 0100
# Means: Do a calculation and give me a result.”

print(x.imag)
print(x.bit_length())
```
