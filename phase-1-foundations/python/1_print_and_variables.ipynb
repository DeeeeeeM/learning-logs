# Prints and Variables
Shows how to display values with print() and how to use variables.

```python
# print() displays text or values to the console.
print("Hello, World!")
```

```python
# Variables store values so we can reuse them later.
# Python automatically detects the type of the value.
name = "Heidel"        # string - a sequence of characters
age = 30                # integer - a whole number
gpa = 3.5               # float - a number with a decimal point
is_student = True       # boolean - True or False value
```

```python
# Use f-strings to create readable output with embedded values.
print(f"Name: {name}")
print(f"Age: {age}")
print(f"GPA: {gpa}")
print(f"Is student?: {is_student}")
```

```python
# Variables can be updated:
age = age + 1
print(f"One year later, age = {age}")
```

```python
# A variable can also hold a new type of value.
name = 123
print(f"Now name holds a number: {name} ({type(name).__name__})")
```

```python
# Strings can be multiplied using * operator
# technical term for this is "operator overloading"
student = "Spam" * 4
```

```python
# print function has several optional arguments

help(print)

print(1, 2, 3, sep=' < ') # separates the strings with "<"" instead of single spaces

#==============================#

def greet(who="Colin"):
    print("Hello,", who)
    
greet()
greet(who="Kaggle")

# We don't need to specify "who" because it's unambiguous, it has a default value and is optional

```
